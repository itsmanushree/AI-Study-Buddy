## 🛠️ Step 5 — Give Our Study Buddy Some Tools

A normal LLM mainly generates responses.

An AI agent can be given **tools** that it can choose to use.

We will create three Python tools:

1. 📖 `explain_concept()` — explains a topic
2. ❓ `generate_quiz()` — creates a quiz
3. 📝 `make_revision_notes()` — creates revision notes

Later, Gemini will decide which tool is appropriate for the student's request.

```python
def explain_concept(topic: str) -> str:
    """Explain a computer science concept to a beginner."""
    prompt = f"""
You are an encouraging computer science tutor.

Explain the concept '{topic}' to a beginner.

Include:
1. A simple definition
2. A real-life analogy
3. 3 key points
4. One simple example

Keep the explanation clear and concise.
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text
```

```python
print(explain_concept("Recursion"))
```

```python
def generate_quiz(topic: str) -> str:
    """Create a short beginner-friendly quiz about a topic."""
    prompt = f"""
You are a helpful computer science tutor.

Create a short quiz on '{topic}' for a beginner.

Generate:
1. 3 multiple-choice questions
2. 4 options for each question (A, B, C, D)
3. The correct answer after each question
4. A one-line explanation of the correct answer

Keep the quiz clear and concise.
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text
```

```python
print(generate_quiz("Recursion"))
```

```python
def make_revision_notes(topic: str) -> str:
    """Create concise revision notes about a topic."""
    prompt = f"""
You are a computer science tutor helping a beginner revise.

Create concise revision notes for '{topic}'.

Include:
1. What it is
2. Important concepts
3. Key points to remember
4. One simple example
5. Common mistake students make

Use clear headings and bullet points.
Keep it under 200 words.
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text
```

## 🤖 Step 6 — Turn Our Tools Into an AI Agent

We now give Gemini access to our three Python functions.

Gemini does not simply follow a hard-coded `if/else` decision.

Instead, it receives the student's request and decides which tool would best help.

For example:

Student: "Teach me recursion."
→ Gemini chooses `explain_concept`

Student: "Quiz me on recursion."
→ Gemini chooses `generate_quiz`

Student: "Give me revision notes on recursion."
→ Gemini chooses `make_revision_notes`

This is the key idea behind **tool calling**.

```python
from google.genai import types

explain_declaration = {
    "name": "explain_concept",
    "description": "Explains a computer science concept to a beginner.",
    "parameters": {
        "type": "object",
        "properties": {
            "topic": {
                "type": "string",
                "description": "The computer science concept to explain."
            }
        },
        "required": ["topic"]
    }
}

quiz_declaration = {
    "name": "generate_quiz",
    "description": "Creates a short beginner-friendly quiz about a topic.",
    "parameters": {
        "type": "object",
        "properties": {
            "topic": {
                "type": "string",
                "description": "The topic for the quiz."
            }
        },
        "required": ["topic"]
    }
}

revision_declaration = {
    "name": "make_revision_notes",
    "description": "Creates concise revision notes about a topic.",
    "parameters": {
        "type": "object",
        "properties": {
            "topic": {
                "type": "string",
                "description": "The topic for the revision notes."
            }
        },
        "required": ["topic"]
    }
}

tools = types.Tool(
    function_declarations=[
        explain_declaration,
        quiz_declaration,
        revision_declaration
    ]
)

agent_config = types.GenerateContentConfig(
    tools=[tools]
)

print("Study Buddy Agent is ready!")
```

### 🧠 How the Agent Works

Our agent follows this basic workflow:

Student Request
↓
Gemini understands the request
↓
Gemini selects a tool
↓
Python executes that tool
↓
Study Buddy returns the result

We also keep a small conversation history so the agent can understand references such as "it", "this topic", or "the same thing".

```python
conversation_history = []


def study_buddy(user_request):

    global last_topic

    # Give Gemini the previous conversation for context
    context = ""

    if conversation_history:
        context = "\nPrevious conversation:\n"
        for user_msg, assistant_msg in conversation_history[-4:]:
            context += f"Student: {user_msg}\n"
            context += f"Study Buddy: {assistant_msg}\n"

    prompt = f"""
You are an AI Study Buddy for a computer science student.

Your job is to understand what the student wants and use the appropriate tool.

Available tools:
- explain_concept: explain a concept
- generate_quiz: create a quiz
- make_revision_notes: create revision notes

Important:
- Use the previous conversation to understand references such as
  "it", "this", "that topic", or "the same thing".
- If the student says "quiz me on it", use the most recently discussed
  computer science topic.
- If the student says "make revision notes", use the most recently
  discussed topic unless they specify another one.

{context}

Current student request:
{user_request}
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config=agent_config
    )

    function_call = response.candidates[0].content.parts[0].function_call

    if function_call:

        tool_name = function_call.name
        args = function_call.args

        print("Agent selected tool:", tool_name)
        print("Arguments:", args)

        if tool_name == "explain_concept":
            result = explain_concept(**args)

        elif tool_name == "generate_quiz":
            result = generate_quiz(**args)

        elif tool_name == "make_revision_notes":
            result = make_revision_notes(**args)

        else:
            result = "Unknown tool selected."

        # Save conversation
        conversation_history.append((user_request, result))

        return result

    else:
        result = response.text
        conversation_history.append((user_request, result))
        return result
```

```python
print("🎓 AI Study Buddy")
print("Ask anything about your studies.")
print("Type 'exit' when you're done.\n")

while True:
    user_request = input("You: ").strip()

    if user_request.lower() == "exit":
        print("👋 Study session ended!")
        break

    if not user_request:
        continue

    print("\n🤖 Study Buddy:")
    print(study_buddy(user_request))
    print("\n" + "-" * 50 + "\n")
```
